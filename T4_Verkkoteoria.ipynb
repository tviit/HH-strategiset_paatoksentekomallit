## Te, Tl ja pelivarat

#### Te-sarake

Lisätään tietokehykseen jokaista työvaihetta vastaava ensimmäistä valmistumisaikaa vastaava sarake siten, että jokaista työvaihetta kohti käydään läpi edellisen vaiheen ensimmäinen valmistumisaika, johon aina lisätään tähän työvaiheeseen tarvittava aika. Niissä työvaiheissa, johon voidaan päätyä useampaa reittiä edellisen työvaiheen valmistumisen ajankohdaksi otetaan näistä viimeisenä valmistuva käyttäen max-funktiota. Käytetään tästä uudesta sarakkeesta nimeä "Te".

```python
# Te-sarake.

remontti.loc[0, 'Te'] = remontti.loc[0, 'Päivää']
remontti.loc[1, 'Te'] = remontti.loc[0, 'Te'] + remontti.loc[1, 'Päivää']
remontti.loc[2, 'Te'] = remontti.loc[1, 'Te'] + remontti.loc[2, 'Päivää']
remontti.loc[3, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[3, 'Päivää']
remontti.loc[4, 'Te'] = max(remontti.loc[3, 'Te'], remontti.loc[9, 'Te']) + remontti.loc[4, 'Päivää']
remontti.loc[5, 'Te'] = remontti.loc[4, 'Te'] + remontti.loc[5, 'Päivää']
remontti.loc[6, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[6, 'Päivää']
remontti.loc[7, 'Te'] = remontti.loc[6, 'Te'] + remontti.loc[7, 'Päivää']
remontti.loc[8, 'Te'] = remontti.loc[7, 'Te'] + remontti.loc[8, 'Päivää']
remontti.loc[9, 'Te'] = remontti.loc[8, 'Te'] + remontti.loc[9, 'Päivää']
remontti.loc[10, 'Te'] = remontti.loc[6, 'Te'] + remontti.loc[10, 'Päivää']
remontti.loc[11, 'Te'] = remontti.loc[2, 'Te'] + remontti.loc[11, 'Päivää']
remontti.loc[12, 'Te'] = remontti.loc[11, 'Te'] + remontti.loc[12, 'Päivää']
remontti.loc[13, 'Te'] = max(remontti.loc[6, 'Te'], remontti.loc[12, 'Te']) + remontti.loc[13, 'Päivää']
remontti.loc[14, 'Te'] = remontti.loc[11, 'Te'] + remontti.loc[14, 'Päivää']
remontti.loc[15, 'Te'] = max(remontti.loc[10, 'Te'], remontti.loc[13, 'Te'], remontti.loc[14, 'Te']) + remontti.loc[15, 'Päivää']
remontti.loc[16, 'Te'] = remontti.loc[15, 'Te'] + remontti.loc[16, 'Päivää']
remontti.loc[17, 'Te'] = remontti.loc[16, 'Te'] + remontti.loc[17, 'Päivää']
remontti.loc[18, 'Te'] = remontti.loc[16, 'Te'] + remontti.loc[18, 'Päivää']
remontti.loc[19, 'Te'] = remontti.loc[16, 'Te'] + remontti.loc[19, 'Päivää']
remontti.loc[20, 'Te'] = max(remontti.loc[18, 'Te'], remontti.loc[19, 'Te']) + remontti.loc[20, 'Päivää']
remontti.loc[21, 'Te'] = max(remontti.loc[17, 'Te'], remontti.loc[20, 'Te']) + remontti.loc[21, 'Päivää']
remontti.loc[22, 'Te'] = remontti.loc[20, 'Te'] + remontti.loc[22, 'Päivää']
remontti.loc[23, 'Te'] = max(remontti.loc[5, 'Te'], remontti.loc[21, 'Te'], remontti.loc[22, 'Te']) + remontti.loc[23, 'Päivää']

remontti
```

#### Tl-sarake

Lisätään sarake, jolla seurataan viimeistä ajankohtaa, jolloin kunkin toiminnon olisi viimeistään oltava valmis. Nyt edetään päinvastaisessa järjestyksessä siten, että aloitusriville tulee edellä määritellyn 'Te'-sarakkeen viimeinen arvo. Tästä suunasta edettäessä niitä työvaiheita kohti, joissa joihin voidaan vastaavasti päätyä useampaa reittiä, otetaan alkupisteeksi edellisten vaiheiden minimi min-funktiolla. Käytetään tästä sarakkeesta nimeä 'Tl'.

Havaitaan myös, että tällä tavalla tietokehystä luotaessa tulee varmistaa, ettei rivejä tule lisätyksi väärässä järjestyksessä ajaen, sillä muuten saataisiin "NaN"-rivejä. Tästä syystä käsittelyjärjestystä on seuraavassa hieman hienosäädetty niin, että tätä ei tapahtuisi.

```python
# Tl-sarake.

remontti.loc[23, 'Tl'] = remontti.loc[23, 'Te']
remontti.loc[22, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Päivää']
remontti.loc[21, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Päivää']
remontti.loc[20, 'Tl'] = min(remontti.loc[21, 'Tl'] - remontti.loc[21, 'Päivää'], remontti.loc[22, 'Tl'] - remontti.loc[22, 'Päivää'])
remontti.loc[19, 'Tl'] = remontti.loc[20, 'Tl'] - remontti.loc[20, 'Päivää']
remontti.loc[18, 'Tl'] = remontti.loc[20, 'Tl'] - remontti.loc[20, 'Päivää']
remontti.loc[17, 'Tl'] = remontti.loc[21, 'Tl'] - remontti.loc[21, 'Päivää']
remontti.loc[16, 'Tl'] = min(remontti.loc[17, 'Tl'] - remontti.loc[17, 'Päivää'], remontti.loc[18, 'Tl'] - remontti.loc[18, 'Päivää'], remontti.loc[19, 'Tl'] - remontti.loc[19, 'Päivää'])
remontti.loc[15, 'Tl'] = remontti.loc[16, 'Tl'] - remontti.loc[16, 'Päivää']
remontti.loc[14, 'Tl'] = remontti.loc[15, 'Tl'] - remontti.loc[15, 'Päivää']
remontti.loc[13, 'Tl'] = remontti.loc[15, 'Tl'] - remontti.loc[15, 'Päivää']
remontti.loc[12, 'Tl'] = remontti.loc[13, 'Tl'] - remontti.loc[13, 'Päivää']
remontti.loc[11, 'Tl'] = min(remontti.loc[12, 'Tl'] - remontti.loc[12, 'Päivää'], remontti.loc[14, 'Tl'] - remontti.loc[14, 'Päivää'])
remontti.loc[10, 'Tl'] = remontti.loc[15, 'Tl'] - remontti.loc[15, 'Päivää']
remontti.loc[5, 'Tl'] = remontti.loc[23, 'Tl'] - remontti.loc[23, 'Päivää']
remontti.loc[4, 'Tl'] = remontti.loc[5, 'Tl'] - remontti.loc[5, 'Päivää']
remontti.loc[9, 'Tl'] = remontti.loc[4, 'Tl'] - remontti.loc[4, 'Päivää']
remontti.loc[8, 'Tl'] = remontti.loc[9, 'Tl'] - remontti.loc[9, 'Päivää']
remontti.loc[7, 'Tl'] = remontti.loc[8, 'Tl'] - remontti.loc[8, 'Päivää']
remontti.loc[6, 'Tl'] = min(remontti.loc[7, 'Tl'] - remontti.loc[7, 'Päivää'], remontti.loc[10, 'Tl'] - remontti.loc[10, 'Päivää'], remontti.loc[13, 'Tl'] - remontti.loc[13, 'Päivää'])
remontti.loc[3, 'Tl'] = remontti.loc[4, 'Tl'] - remontti.loc[4, 'Päivää']
remontti.loc[2, 'Tl'] = min(remontti.loc[3, 'Tl'] - remontti.loc[3, 'Päivää'], remontti.loc[6, 'Tl'] - remontti.loc[6, 'Päivää'], remontti.loc[11, 'Tl'] - remontti.loc[11, 'Päivää'])
remontti.loc[1, 'Tl'] = remontti.loc[2, 'Tl'] - remontti.loc[2, 'Päivää']
remontti.loc[0, 'Tl'] = remontti.loc[1, 'Tl'] - remontti.loc[1, 'Päivää']


remontti
```

Sisällytän nämä lähinnä kuriositeettina ja myöhempää lähempää tutustumista varten. Koska nämä eivät nähdäkseni kuulu tehtävänantoon, näiden toimintaa ei varsinaisesti myöskään tarkastella tätä enemmin.

```python
class BellmanFord:
    def __init__(self, nodes):
        self.nodes = nodes
        self.edges = []
        
    def add_edge(self, node_a, node_b, weight):
        self.edges.append((node_a, node_b, weight))
    
    def shortest_path(self, start_node, end_node):
        distances = {}
        for node in self.nodes:
            distances[node] = float("inf")
        distances[start_node] = 0
        previous = {}
        previous[start_node] = None
        
        for _ in range(len(self.nodes) - 1):
            for edge in self.edges:
                node_a, node_b, weight = edge
                new_distance = distances[node_a] + weight
                if new_distance < distances[node_b]:
                    distances[node_b] = new_distance
                    previous[node_b] = node_a
                    
        if distances[end_node] == float("inf"):
            return None
          
        path = []
        node = end_node
        while node:
            path.append(node)
            node = previous[node]
            
        path.reverse()
        return path
```

```python
class FloydWarshall:
    def __init__(self, nodes):
        self.nodes = nodes
        self.graph = {}
        for a in self.nodes:
            for b in self.nodes:
                distance = 0 if a == b else float("inf")
                self.graph[(a, b)] = distance

    def add_edge(self, a, b, w):
        self.graph[(a, b)] = min(self.graph[(a, b)], w)
        
    def find_distances(self):
        distances = self.graph.copy()

        for k in self.nodes:
            for a in self.nodes:
                for b in self.nodes:
                    distance = min(distances[(a, b)],
                                   distances[(a, k)] + 
                                   distances[(k, b)])
                    distances[(a, b)] = distance
        return distances
```
